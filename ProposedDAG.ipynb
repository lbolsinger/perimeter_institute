```python
import semopy
import pandas as pd

desc = '''
# structure
retention ~ financial_elasticity + brand_association + opportunity_cost + loyalty + proposed_rate_change

financial_elasticity ~ disposible_income + budget_consciousness + price_sensitivity + value_perception
brand_association ~ duration + affinity + public_image
opportunity_cost ~ competition_awareness + decision_fatigue + switching_costs
loyalty ~ risk_aversion + default_bias + trust + inertia

disposible_income ~ income + expenses
expenses ~ vehicle_type + postal_code + credit_score + inflation
budget_consciousness ~ coverage + previous_premium + joint_plans
price_sensitivity ~ proposed_rate_change + effective_rate_change + expense_change
effective_rate_change ~ inflation + previous_premium + proposed_rate_change
expense_change ~ inflation + change_postal_code + change_vehicle_type
value_perception ~ coverage + claim_history + vehicle_usage + vehicle_type + joint_plans

duration ~ total_years + consecutive_years + number_of_policies
affinity ~ claim_history
public_image ~ postal_code + age + gender

competition_awareness ~ age + gender + postal_code + income + broker + consecutive_years
decision_fatigue ~ age + marital_status + number_of_policies + time_of_proposed_change + joint_plans + plan_complexity + auto_renewal
switching_costs ~ claims_in_progress + joint_plans + driver_abstract

risk_aversion ~ coverage + age + vehicle_usage + driver_abstract + claim_history + lawyer
default_bias ~ auto_renewal + total_years
trust ~ claim_history
inertia ~ age + consecutive_years + auto_renewal + plan_stability

previous_premium ~ age + marital_status + gender + coverage + joint_plans + driver_abstract + claim_history + inflation + postal_code + vehicle_type + vehicle_usage
proposed_rate_change ~ age + marital_status + gender + coverage + joint_plans + driver_abstract + claim_history + inflation + postal_code + vehicle_type + vehicle_usage

# latent definitions
DEFINE(latent) financial_elasticity disposible_income expenses budget_consciousness price_sensitivity effective_rate_change expense_change value_perception
DEFINE(latent) brand_association duration affinity public_image
DEFINE(latent) opportunity_cost competition_awareness decision_fatigue switching_costs
DEFINE(latent) loyalty risk_aversion default_bias trust inertia

# ordinal definitions
DEFINE(ordinal) vehicle_type plan_stability plan_complexity auto_renewal joint_plans
DEFINE(ordinal) gender marital_status lawyer driver_abstract claims_in_progress claim_history

# fixed variance
financial_elasticity ~~ 1*financial_elasticity
brand_association ~~ 1*brand_association
opportunity_cost ~~ 1*opportunity_cost
loyalty ~~ 1*loyalty
'''

g = semopy.semplot(desc, "model_plot.png")
```
